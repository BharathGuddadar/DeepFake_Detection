Frame Extraction 

```python
import cv2
import os

def extract_frames(video_path, output_folder, max_frames=30):
    """Extract frames from a video file."""
    if not os.path.exists(video_path):
        print(f"Video file not found: {video_path}")
        return
    
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        print(f"Failed to open video: {video_path}")
        return
    
    os.makedirs(output_folder, exist_ok=True)
    
    count = 0
    while count < max_frames:
        success, frame = cap.read()
        if not success:
            break  # End of video or error reading frame
            
        frame_path = os.path.join(output_folder, f"frame{count:04d}.jpg")
        cv2.imwrite(frame_path, frame)
        count += 1
    
    cap.release()
    print(f"Extracted {count} frames from {video_path}")

def process_all_videos(input_folder, output_folder, label, max_videos=100):
    """Process all videos in a directory."""
    video_dir = os.path.join(input_folder, label)
    if not os.path.exists(video_dir):
        print(f"❌ Folder not found: {video_dir}")
        return

    output_label_dir = os.path.join(output_folder, label)
    os.makedirs(output_label_dir, exist_ok=True)

    count = 0
    for file in os.listdir(video_dir):
        if count >= max_videos:
            break
            
        if file.lower().endswith(".mp4"):
            video_name = file[:-4]  # Remove .mp4 extension
            video_path = os.path.join(video_dir, file)
            frames_output_dir = os.path.join(output_label_dir, video_name)
            
            extract_frames(video_path, frames_output_dir)
            count += 1

    print(f"Processed {count} videos for label '{label}'")

# Process videos
process_all_videos("Dataset", "frames", "real", max_videos=100)
process_all_videos("Dataset", "frames", "fake", max_videos=100)
```
